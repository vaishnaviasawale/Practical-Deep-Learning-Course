```python
from fastbook import *
from fastai.vision.widgets import *
```

```python
btn_upload = widgets.FileUpload()
btn_upload
```

```python
# For the book, we can't actually click an upload button, so we fake it
btn_upload = SimpleNamespace(data = ['images/grizzly.jpg'])
```

```python
img = PILImage.create(btn_upload.data[-1])
```

```python
out_pl = widgets.Output()
out_pl.clear_output()
with out_pl: display(img.to_thumb(128,128))
out_pl
```

```python
path = Path()
path.ls(file_exts='.pkl')
learn_inf = load_learner(path/'export.pkl')
```

```python
# Make a prediction
pred,pred_idx,probs = learn_inf.predict(img)
```

```python
# Use a Label to display them
lbl_pred = widgets.Label()
lbl_pred.value = f'Prediction: {pred}; Probability: {probs[pred_idx]:.04f}'
lbl_pred
     
Label(value='Prediction: grizzly; Probability: 1.0000')
```

```python
# Button to do the classification
btn_run = widgets.Button(description='Classify')
btn_run
```

```python
# A click event handler: a function that will be called when it's pressed. We can 
# just copy over the lines of code from above:
def on_click_classify(change):
    img = PILImage.create(btn_upload.data[-1])
    out_pl.clear_output()
    with out_pl: display(img.to_thumb(128,128))
    pred,pred_idx,probs = learn_inf.predict(img)
    lbl_pred.value = f'Prediction: {pred}; Probability: {probs[pred_idx]:.04f}'
```

```python
btn_run.on_click(on_click_classify)
# Pressing the button should update the image and predictions automatically!
```

```python
# We now put them all in a vertical box (VBox) to complete our GUI:
#Putting back btn_upload to a widget for next cell
btn_upload = widgets.FileUpload()
```

```python
VBox([widgets.Label('Select your bear!'), 
      btn_upload, btn_run, out_pl, lbl_pred])
```
